# While Loop:
      - While Loop runs till the Boolean is True

```python
for age in range(1,18):
    print(f"You are a kid! {age}")   
```

```python
age = 1
while age<18:
    print(f"You are a kid! {age}")   
    age+=1
```

```python
names=['usman', 'phil','arham','amir', 'sarah', 'rehan','taha', 'nasir']
for n in names:
    print(f"Welcome {n}")
```

```python
names=['usman', 'phil','arham','amir', 'sarah', 'rehan', 'nasir']
index =0
while index<8:
    print(f"Welcome {names[index]}")
    index+=1
```

```python
names=['usman', 'phil','arham','amir', 'sarah', 'rehan', 'nasir']
index =0
while index<len(names):
    print(f"Welcome {names[index]}")
    index+=1
```

# Store
     - shoppping 
     - billing

```python
store={ 'sugar':{'price': 200, 'quantity': 1000},
        'rice': {'price': 400, 'quantity': 500},
        'oil': {'price': 600, 'quantity': 2000},
        'meat': {'price': 2000, 'quantity': 1000},
        'flour': {'price': 180, 'quantity': 5000},
        'milk': {'price': 240, 'quantity': 100}
}
```

```python
item = input("Kia chahaye?").lower()
for item_name in store:
    if item_name==item:
        print(f"{item } Avaiable hy") 
        print(f"{store[item_name]['price']} rupey per KG hy!")
        kgs = int(input('How many kGs'))
        if kgs<=store[item_name]['quantity']:
            bill = store[item_name]['price'] * kgs
            store[item_name]['quantity']-=kgs
            print(f"Please pay amout {bill}")
        else:
            print(f"{kgs} Kgs {item_name} not available ")
        
        break 
else:
    print(f"{item } not Avaiable")   
```
