# Nested Conditons
  - One inside another

```python
python = int(input("Enter python marks"))
numpy = int(input("Enter numpy marks"))
pandas  = int(input("Enter pandas marks"))
stats = int(input("Enter stats marks"))
mL = int(input("Enter mL marks"))
total = python + numpy + pandas + stats + mL
percentage  = total / 500 * 100
result  = None
grade = None
if python>=50:
    if numpy>=50:
        if pandas>=50:
            if stats>=50:
                if mL>=50:
                    result = "Pass"
                else:
                    result ='Fail'
            else:
                result ='Fail'
        else:
            result ='Fail'
    else:
        result ='Fail'
else:
    result ='Fail'
if result=="Pass":
    if percentage>=90:
        grade  = "A+"
    elif percentage>=80:
        grade = "A"
    elif percentage>=70:
        grade = "B"
    elif percentage>=60:
        grade = "C"
    elif percentage>=50:
        grade = "D"
    else:
        grade = "F"
print("Total: ", total)
print("Percentage: ", percentage)
print("Result:", result)
print("Grade: ", grade)
```

```python
guests = ['nasir', 'tahir', 'asad', 'usman', 'ali', 'faisal', 'hassan']
```

```python
for name in guests:
    print(name)
```

```python
for name in guests:
    print(f"Dear Mr.{name.title()}. You are invited for a dinner tomorrow at my house.")
```

```python
for name in guests:
    if not name.startswith('a'):
        print(f"Dear Mr.{name.title()}. You are invited for a dinner tomorrow at my house.")
```

```python
for name in guests:
    if name.startswith('a')==False:
        print(f"Dear Mr.{name.title()}. You are invited for a dinner tomorrow at my house.")
```

```python
for name in guests:
    if name.startswith('a'):
        pass        
    else:
        print(f"Dear Mr.{name.title()}. You are invited for a dinner tomorrow at my house.")
```

```python
numbers  = [1,4,23,43,5,67,7,87,9,21,43,76,90,92,56,51,62,95]
```

```python
for n in numbers:
    if n%2==0:
        print(f"Number { n } is an even number")
    else:
        print(f"Number { n } is an odd number")
```

```python
evens = []
odds  = []
for n in numbers:
    if n%2==0:
        evens.append(n)
    else:
        odds.append(n)
print(evens)
odds
```
