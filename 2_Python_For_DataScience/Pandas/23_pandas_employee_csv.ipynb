```python
#  Finding missing data
data.isna().sum()
```

```python
data.dropna(how='any')
# it drops rows by default (axis=0)
# it drops all the row where any of the cell is na.
```

```python
data.dropna(how='all')
# it drops rows by default (axis=0)
# it drops all the row where any of the cell is na.
```

```python
data.dropna(subset=['Salary','Designation'],how='any')
# it drops rows by default (axis=0)
# it drops all the row where Salary or Designation is missing
```

```python
data.dropna(subset=['Salary','Designation'],how='all')
# it drops rows by default (axis=0)
# it drops all the row where Salary and Designation are  missing together.
```

```python
data.dropna(thresh=int(data.shape[0]*0.90),axis=1)
# jitna bhi data hy usme agar 24 values thk hen to column delte nahi hoga
```

```python
data.dropna(thresh=5,axis=0)
# jitna bhi data hy usme agar 5 values record me thk thk hen to record delte nahi hoga
```
