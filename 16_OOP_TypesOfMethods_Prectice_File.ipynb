# Types of Methods

- class methods 
- instance/object methods
- static methods

```python
class Student:    
    #class variable
    school = "Usman Institue of Technology"    
    def __init__(self,m1,m2,m3):
        #instance variable
        self.m1=m1
        self.m2=m2
        self.m3= m3   
    # instance methods    
    def avg(self):
        return (self.m1+self.m2+self.m3)/3    
    @classmethod
    def getSchool(cls):    
        return cls.school  

    @staticmethod
    def info():
        print("This is student class dear") 
```

```python
s1 = Student(89,98,99) 
s2 = Student(98,98,99) 
```

```python
#Calling an instance method
print(s1.avg())
```

```python
print(Student.avg()) # class can not call instance method wthout object/instance
```

```python
# Calling a class method Class.
print(Student.getSchool())
```

```python
# Calling a class method via objct
print(s1.getSchool())
```

```python
# Class is calling a static method. OK
Student.info()
```

```python
# Object is calling a static methd. OK
s1.info()
```

# Object/ Instance as attribute

```python
class Car():
   
    """A simple attempt to represent a car."""
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0
    
    def get_descriptive_name(self):
        long_name = str(self.year) + ' ' + self.make + ' ' + self.model
        return long_name.title()
    
    def read_odometer(self):
        print("This car has " + str(self.odometer_reading) + " miles on it.")
    
    def update_odometer(self, mileage):
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            return("You can't roll back an odometer!")
    
    def increment_odometer(self, miles):
        self.odometer_reading += miles
    
    def fill_gas_tank(self):    
        return ("Gas Tank is full")
```

```python
class Battery:
    def __init__(self, size, amp, volts, price, manuf):
        self.size = size
        self.amp =amp
        self.volts = volts
        self.price = price
        self.manuf = manuf
    def descBattery(self):
        print(f"Price:{self.price}\nAmp:{self.amp}\nVolts:{self.volts}\nSize:{self.size} ")
```

```python
b1  = Battery('12inch', 100, 12, 20000, 'Osaka')
```

```python
class ElectricCar(Car):
    """"Represent aspects of a car, specific to electric vehicles."""
    def __init__(self, make, model, year):
        
        '''Initialize attributes of the parent class.'''
        super().__init__(make, model, year)
        self.battery = Battery('12inch', 100, 12, 20000, 'Osaka')
```

```python
my_tesla = ElectricCar('tesla', 'model s', 2016)
```

```python
my_tesla.year
```

```python
my_tesla.battery.descBattery()
```
