# Adding more members to a list
    - append('member'): add at very last index
    - insert(index, member): adds at given index
    - extend
    - + 

```python
courses.append('graphics designing')
courses
```

```python
courses.insert(3, 'cloudcomputing')
courses
```

```python
courses.insert(13, 'javascript')
courses
```

```python
purane_courses = ['typing', 'msoffice','dos']
```

```python
courses.append(purane_courses)
courses
```

```python
courses.extend(purane_courses)
```

```python
courses
```
