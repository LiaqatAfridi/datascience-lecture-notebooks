```python
from  abc import ABC , abstractmethod
```

```python
#abstraction
```

```python
class TVRemote(ABC):  # abstract class ka object ni bnta
  @abstractmethod   # decoder
  def power_button(self):
    pass
  @abstractmethod  # decoder
  def volume(self):
    pass
    # agar mai kesi method ya function ko abstract bana rha hon to mujhe neche define krna pare ga
class Realremote(TVRemote):
  def power_button(self):
    print("TV on ho gaya")
  def volume(self):
    print('real remote ka ')
```

```python
c1 = Realremote()
c1.power_button()
```

```python
c1.volume()
```
