```python
price  = int(input("What is the price of 1KG tomato"))
if price>250:
    print('1KG')
elif price>200:
    print('2KG')
elif price>150:
    print('3KG')
elif price>100:
    print('4KG')
else:
    print('5KG')
```

```python
python = int(input("Enter python marks"))
numpy = int(input("Enter numpy marks"))
pandas  = int(input("Enter pandas marks"))
stats = int(input("Enter stats marks"))
mL = int(input("Enter mL marks"))
total = python + numpy + pandas + stats + mL
percentage  = total / 500 * 100
if percentage>=90:
    grade  = "A+"
elif percentage>=80:
    grade = "A"
elif percentage>=70:
    grade = "B"
elif percentage>=60:
    grade = "C"
elif percentage>=50:
    grade = "D"
else:
    grade = "F"
print("Total: ", total)
print("Percentage: ", percentage)
print("Grade: ", grade)
```

```python
items = {'laptop':50000, 'mobile':60000, 'smartwatch':10000,'earbuds':5000}
```

```python
item_name = input("Enter the item you want to buy")
if item_name in items:
    print(f"The price of {item_name} is: {items.get(item_name)}")
else:
    print(f"The {item_name} is not available at our store")
```

```python
items['smartwatch']
```

```python
item1 = input("Please enter item name to buy")
item2 = input("Please enter item name to buy")
item3 = input("Please enter item name to buy")
bill_amt = items[item1] + items[item2]  + items[item3]
if bill_amt>50000:
    discount = bill_amt*.10
elif bill_amt>25000:
    discount = bill_amt*0.05
else:
    discount = bill_amt*0
amount_payable = bill_amt - discount

print(f"""
      
    {item1} : {items[item1]}
    {item2} : {items[item2]}
    {item3} : {items[item3]}

    Total : {bill_amt}
    Discount: {discount}

    Total Paybale: {amount_payable}

    Thank you for shopping!

""")
```
