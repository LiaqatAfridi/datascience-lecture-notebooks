```python
ek child class 2 ya 2 sau zyada parent class se inherit le
```

```python
class Father:
  def father_skill(self):
    print("father  :     businnes maan" )

class Mother:
  def mother_skill(self):
    print("Mother  :    Cookiing")   # parent class

class child(Father, Mother):   # child classs
  def child_skill(self):
    print('child  :   Coding')
```

```python
c1 = child()
c1.father_skill()
c1.mother_skill()
c1.child_skill()
```

```python
class Employe:  # parent class 1
  company  = "Saylani "  # class variables

  def employ_info(self):
    print("I ama an employee")

  def show_info(self):  # methods ya function
    print(f"company:   My company name is {self.company}")  # behaviour
```

```python
class AiEnginner:# parent class 2
  def ai_work(self):# methods
    print("I build Ai models")
  def train_model(self): # method
    print("Khud ka gpt banaya")
```

```python
class AIemployyy(Employe, AiEnginner):  # child class nai 2 parent class ko inherit kia
  def __init__(self, name, age):  # initializer/ contstroctor
    self.name = name  # instance variable
    self.age = age
  def show_info(self): # methid
    print(f"Name    :    {self.name}")
    print(f"Age     :  {self.age}")
    print(f"company : {self.company}")
  def Employe_parent_show_info(self):  # creat new fnction jo parent class kai method ko call kr la value lai ga
    super().show_info()
```

```python
ai = AIemployyy("Usman", 20)
```

```python
ai.show_info()
```

```python
ai.Employe_parent_show_info()
```
